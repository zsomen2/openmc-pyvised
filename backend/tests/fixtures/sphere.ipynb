```python
%%time
water = openmc.Material(material_id=3, name="water")
water.add_nuclide("H1", 2.0)
water.add_nuclide("O16", 1.0)
water.set_density("g/cm3", 1.0)
sphere = openmc.Sphere(surface_id=1, r=5.0, boundary_type="vacuum")
cell = openmc.Cell(cell_id=7, name="water sphere", fill=water, region=-sphere)
```

```python
model = openmc.Model(geometry=openmc.Geometry([cell]), materials=openmc.Materials([water]))
model.settings.run_mode = "fixed source"
model.settings.particles = 10
model.settings.batches = 1
if True:
    %time model.export_to_model_xml()
display(model)
model.run()
```
